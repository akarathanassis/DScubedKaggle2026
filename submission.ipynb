```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
train = pd.read_csv('train.csv', index_col='ID')
test = pd.read_csv('test.csv', index_col='ID')
```

```python
price_cols = [f"p{i}" for i in range(1, 41)]

ax = test[price_cols].T.plot(
    figsize=(12, 6),
    legend=False,
    linewidth=0.4,
    alpha=0.08,
)
ax.set_xlabel("Time")
ax.set_ylabel("Price")
ax.set_title("Test Stock Price Series")
ax.set_xticks(range(len(price_cols)))
ax.set_xticklabels(price_cols, rotation=45)
plt.tight_layout()
```

```python
test["signal"] = test["p40"] - 100

top_4000 = (
    test[(test["signal"] < 0) & (test["p40"] != 0)]
    .sort_values("p40", ascending=False)
    .head(4000)
)

test["sell"] = 0
test.loc[top_4000.index, "sell"] = 1
submission = test[["sell"]]
submission.to_csv("submission.csv")
```
